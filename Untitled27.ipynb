```python
import os, json, glob, random
import numpy as np
import nibabel as nib
import matplotlib.pyplot as plt
import torch
import monai
from monai.data import Dataset, DataLoader
from monai.transforms import (
    Compose, LoadImaged, EnsureChannelFirstd, Orientationd, Spacingd,
    ScaleIntensityRanged, Lambdad, SpatialPadd, RandCropByPosNegLabeld,
    RandFlipd, RandRotate90d, RandShiftIntensityd, EnsureTyped,
)
from monai.utils import set_determinism

from google.colab import drive
drive.mount('/content/drive')

print("MONAI:", monai.__version__, "| GPU:", torch.cuda.is_available())
```

```python
ROOT = "/content/drive/MyDrive/liver_vessel_unet/data"
DATA_DIR = os.path.join(ROOT, "Task08_HepaticVessel")

SEED = 0
VESSEL_LABEL = 1                 # checked against dataset.json below
PIXDIM = (1.5, 1.5, 2.0)         # target voxel spacing in mm
HU_MIN, HU_MAX = -100, 250       # CT window used for the visualisation in step 1
PATCH = (96, 96, 96)             # 3D patch size (reduce to (64,64,64) if the GPU runs out of memory)
VAL_FRACTION = 0.2
SUBSET_N = None                  # e.g. 40 to debug quickly with only 40 cases; None = all cases

set_determinism(seed=SEED)
```

```python
with open(os.path.join(DATA_DIR, "dataset.json")) as f:
    info = json.load(f)
print("Labels:", info["labels"])
assert "vessel" in info["labels"][str(VESSEL_LABEL)].lower(), \
    "Label 1 is not the vessel: change VESSEL_LABEL above"
print("OK: label", VESSEL_LABEL, "=", info["labels"][str(VESSEL_LABEL)])
```

```python
images = sorted(p for p in glob.glob(os.path.join(DATA_DIR, "imagesTr", "*.nii.gz"))
                if not os.path.basename(p).startswith("._"))
labels = sorted(p for p in glob.glob(os.path.join(DATA_DIR, "labelsTr", "*.nii.gz"))
                if not os.path.basename(p).startswith("._"))
assert len(images) == len(labels), "images and labels do not match"

files = [{"image": i, "label": l} for i, l in zip(images, labels)]
random.Random(SEED).shuffle(files)
if SUBSET_N:
    files = files[:SUBSET_N]

n_val = max(1, int(len(files) * VAL_FRACTION))
val_files, train_files = files[:n_val], files[n_val:]
print(f"Total: {len(files)} | train: {len(train_files)} | validation: {len(val_files)}")

# save the split so that every later step (and your README) uses exactly the same one
split_path = os.path.join(os.path.dirname(ROOT), "split.json")
with open(split_path, "w") as f:
    json.dump({"train": train_files, "val": val_files, "seed": SEED}, f, indent=1)
print("Split saved to", split_path)
```

```python
base = [
    LoadImaged(keys=["image", "label"]),
    EnsureChannelFirstd(keys=["image", "label"]),
    Orientationd(keys=["image", "label"], axcodes="RAS"),
    Spacingd(keys=["image", "label"], pixdim=PIXDIM, mode=("bilinear", "nearest")),
    ScaleIntensityRanged(keys="image", a_min=HU_MIN, a_max=HU_MAX, b_min=0.0, b_max=1.0, clip=True),
    # binary mask: vessel = 1, everything else (background, tumour) = 0
    Lambdad(keys="label", func=lambda x: (x == VESSEL_LABEL).to(torch.float32)),
    SpatialPadd(keys=["image", "label"], spatial_size=PATCH),
]

train_tf = Compose(base + [
    RandCropByPosNegLabeld(keys=["image", "label"], label_key="label",
                           spatial_size=PATCH, pos=1, neg=1, num_samples=2,
                           image_key="image", image_threshold=0),
    RandFlipd(keys=["image", "label"], prob=0.5, spatial_axis=0),
    RandFlipd(keys=["image", "label"], prob=0.5, spatial_axis=1),
    RandRotate90d(keys=["image", "label"], prob=0.3, max_k=3, spatial_axes=(0, 1)),
    RandShiftIntensityd(keys="image", offsets=0.05, prob=0.3),
    EnsureTyped(keys=["image", "label"]),
])

val_tf = Compose(base + [EnsureTyped(keys=["image", "label"])])
print("Transforms ready")
```

```python
import os, json, time
import numpy as np
import matplotlib.pyplot as plt
import torch
import monai
from monai.data import Dataset, DataLoader, decollate_batch
from monai.transforms import (
    Compose, LoadImaged, EnsureChannelFirstd, Orientationd, Spacingd,
    ScaleIntensityRanged, Lambdad, SpatialPadd, RandCropByPosNegLabeld,
    RandFlipd, RandRotate90d, RandShiftIntensityd, EnsureTyped, AsDiscrete,
)
from monai.networks.nets import UNet
from monai.networks.layers import Norm
from monai.losses import DiceCELoss
from monai.metrics import DiceMetric
from monai.inferers import sliding_window_inference
from monai.utils import set_determinism

from google.colab import drive
drive.mount('/content/drive')

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("MONAI:", monai.__version__, "| device:", device)
```

```python
ROOT = "/content/drive/MyDrive/liver_vessel_unet/data"
PROJECT = os.path.dirname(ROOT)
SPLIT_PATH = os.path.join(PROJECT, "split.json")
CKPT_DIR = os.path.join(PROJECT, "checkpoints")
os.makedirs(CKPT_DIR, exist_ok=True)

SEED = 0
VESSEL_LABEL = 1
PIXDIM = (1.5, 1.5, 2.0)
HU_MIN, HU_MAX = -100, 250
PATCH = (96, 96, 96)             # use (64, 64, 64) if the GPU runs out of memory

# smoke-test settings (they will be increased in step 4)
TRAIN_CASES_LIMIT = 40           # number of training cases used
VAL_CASES = 10                   # number of validation cases used
MAX_EPOCHS = 3
BATCH_SIZE = 2                   # cases per step (each case gives 2 patches)
LR = 1e-3

set_determinism(seed=SEED)
torch.backends.cudnn.benchmark = True
use_amp = torch.cuda.is_available()
```

```python
with open(SPLIT_PATH) as f:
    split = json.load(f)
train_files = split["train"][:TRAIN_CASES_LIMIT]
val_files = split["val"][:VAL_CASES]
print("train cases:", len(train_files), "| validation cases:", len(val_files))

base = [
    LoadImaged(keys=["image", "label"]),
    EnsureChannelFirstd(keys=["image", "label"]),
    Orientationd(keys=["image", "label"], axcodes="RAS"),
    Spacingd(keys=["image", "label"], pixdim=PIXDIM, mode=("bilinear", "nearest")),
    ScaleIntensityRanged(keys="image", a_min=HU_MIN, a_max=HU_MAX, b_min=0.0, b_max=1.0, clip=True),
    Lambdad(keys="label", func=lambda x: (x == VESSEL_LABEL).to(torch.float32)),
    SpatialPadd(keys=["image", "label"], spatial_size=PATCH),
]
train_tf = Compose(base + [
    RandCropByPosNegLabeld(keys=["image", "label"], label_key="label",
                           spatial_size=PATCH, pos=1, neg=1, num_samples=2,
                           image_key="image", image_threshold=0),
    RandFlipd(keys=["image", "label"], prob=0.5, spatial_axis=0),
    RandFlipd(keys=["image", "label"], prob=0.5, spatial_axis=1),
    RandRotate90d(keys=["image", "label"], prob=0.3, max_k=3, spatial_axes=(0, 1)),
    RandShiftIntensityd(keys="image", offsets=0.05, prob=0.3),
    EnsureTyped(keys=["image", "label"]),
])
val_tf = Compose(base + [EnsureTyped(keys=["image", "label"])])

train_loader = DataLoader(Dataset(train_files, train_tf), batch_size=BATCH_SIZE, shuffle=True, num_workers=2)
val_loader = DataLoader(Dataset(val_files, val_tf), batch_size=1, shuffle=False, num_workers=2)
```

```python
model = UNet(
    spatial_dims=3, in_channels=1, out_channels=2,
    channels=(16, 32, 64, 128, 256), strides=(2, 2, 2, 2),
    num_res_units=2, norm=Norm.INSTANCE,
).to(device)

loss_fn = DiceCELoss(to_onehot_y=True, softmax=True, include_background=False)
optimizer = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=1e-5)
scaler = torch.amp.GradScaler("cuda", enabled=use_amp)

n_params = sum(p.numel() for p in model.parameters())
print(f"Parameters: {n_params/1e6:.2f} M")
```

```python
batch = next(iter(train_loader))
x, y = batch["image"].to(device), batch["label"].to(device)
with torch.no_grad():
    out = model(x)
print("input:", tuple(x.shape), "| output:", tuple(out.shape), "| label:", tuple(y.shape))
print("initial loss (expected around 1 to 1.7):", float(loss_fn(out, y)))
```

```python
post_pred = AsDiscrete(argmax=True, to_onehot=2)
post_label = AsDiscrete(to_onehot=2)
dice_metric = DiceMetric(include_background=False, reduction="mean")

def validate():
    model.eval()
    dice_metric.reset()
    with torch.no_grad():
        for vb in val_loader:
            vx, vy = vb["image"].to(device), vb["label"].to(device)
            with torch.autocast(device_type="cuda", dtype=torch.float16, enabled=use_amp):
                vo = sliding_window_inference(vx, PATCH, 2, model, overlap=0.25)
            preds = [post_pred(i) for i in decollate_batch(vo)]
            gts = [post_label(i) for i in decollate_batch(vy)]
            dice_metric(y_pred=preds, y=gts)
    return float(dice_metric.aggregate().item())

history = {"loss": [], "val_dice": []}
best_dice = -1.0
t0 = time.time()

for epoch in range(MAX_EPOCHS):
    model.train()
    losses = []
    for step, b in enumerate(train_loader):
        x, y = b["image"].to(device), b["label"].to(device)
        optimizer.zero_grad(set_to_none=True)
        with torch.autocast(device_type="cuda", dtype=torch.float16, enabled=use_amp):
            loss = loss_fn(model(x), y)
        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()
        losses.append(loss.item())
        if step % 5 == 0:
            print(f"  epoch {epoch+1} step {step+1}/{len(train_loader)} loss {loss.item():.4f}")
    ep_loss = float(np.mean(losses))
    d = validate()
    history["loss"].append(ep_loss)
    history["val_dice"].append(d)
    if d > best_dice:
        best_dice = d
        torch.save(model.state_dict(), os.path.join(CKPT_DIR, "unet_smoke_best.pt"))
    print(f"epoch {epoch+1}/{MAX_EPOCHS} | train loss {ep_loss:.4f} | val Dice {d:.4f} | {(time.time()-t0)/60:.1f} min")

with open(os.path.join(PROJECT, "smoke_test_history.json"), "w") as f:
    json.dump(history, f, indent=1)
print("Best validation Dice:", round(best_dice, 4))
```

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 3.5))
ax[0].plot(range(1, len(history["loss"]) + 1), history["loss"], marker="o")
ax[0].set_title("Training loss"); ax[0].set_xlabel("epoch")
ax[1].plot(range(1, len(history["val_dice"]) + 1), history["val_dice"], marker="o", color="tab:green")
ax[1].set_title("Validation Dice (vessels)"); ax[1].set_xlabel("epoch")
plt.tight_layout(); plt.show()
```

```python
import os, json, time, shutil
import numpy as np
import matplotlib.pyplot as plt
import torch
import monai
from monai.data import PersistentDataset, DataLoader, decollate_batch
from monai.transforms import (
    Compose, LoadImaged, EnsureChannelFirstd, Orientationd, Spacingd,
    ScaleIntensityRanged, Lambdad, SpatialPadd, CastToTyped, RandCropByPosNegLabeld,
    RandFlipd, RandRotate90d, RandShiftIntensityd, EnsureTyped, AsDiscrete,
)
from monai.networks.nets import UNet
from monai.networks.layers import Norm
from monai.losses import DiceCELoss
from monai.metrics import DiceMetric
from monai.inferers import sliding_window_inference
from monai.utils import set_determinism

from google.colab import drive
drive.mount('/content/drive')

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("MONAI:", monai.__version__, "| device:", device)
```

```python
ROOT = "/content/drive/MyDrive/liver_vessel_unet/data"
PROJECT = os.path.dirname(ROOT)
SPLIT_PATH = os.path.join(PROJECT, "split.json")
CKPT_DIR = os.path.join(PROJECT, "checkpoints")
os.makedirs(CKPT_DIR, exist_ok=True)
CKPT_LAST = os.path.join(CKPT_DIR, "unet_last.pt")
CKPT_BEST = os.path.join(CKPT_DIR, "unet_best.pt")

# Cache of the preprocessed volumes: LOCAL disk (fast). Lost if the Colab runtime is deleted,
# in which case it is simply rebuilt (about one slow epoch).
CACHE_DIR = "/content/cache_vessel"
RESET_CACHE = False      # set True if you change PIXDIM / HU window / VESSEL_LABEL (the cache would be stale)
if RESET_CACHE and os.path.exists(CACHE_DIR):
    shutil.rmtree(CACHE_DIR)

SEED = 0
VESSEL_LABEL = 1
PIXDIM = (1.5, 1.5, 2.0)
HU_MIN, HU_MAX = -100, 250
PATCH = (96, 96, 96)             # (64, 64, 64) if the GPU runs out of memory

BATCH_SIZE = 2                   # cases per step (2 patches per case)
LR = 1e-3
EPOCHS_TOTAL = 60                # one epoch = all training cases once
TIME_BUDGET_MIN = 150            # the loop stops cleanly after this many minutes; run again to continue
VAL_EVERY = 5                    # validate every N epochs
VAL_CASES_DURING = 15            # validation cases used during training (all 60 are used at the end)
OVERFIT_EPOCHS = 6

set_determinism(seed=SEED)
torch.backends.cudnn.benchmark = True
use_amp = torch.cuda.is_available()
```

```python
with open(SPLIT_PATH) as f:
    split = json.load(f)
train_files, val_files = split["train"], split["val"]
print("train:", len(train_files), "| validation:", len(val_files))

base = [
    LoadImaged(keys=["image", "label"]),
    EnsureChannelFirstd(keys=["image", "label"]),
    Orientationd(keys=["image", "label"], axcodes="RAS"),
    Spacingd(keys=["image", "label"], pixdim=PIXDIM, mode=("bilinear", "nearest")),
    ScaleIntensityRanged(keys="image", a_min=HU_MIN, a_max=HU_MAX, b_min=0.0, b_max=1.0, clip=True),
    Lambdad(keys="label", func=lambda x: (x == VESSEL_LABEL).to(torch.uint8)),
    SpatialPadd(keys=["image", "label"], spatial_size=PATCH),
    CastToTyped(keys=["image", "label"], dtype=[torch.float16, torch.uint8]),   # halves the cache size
]
train_tf = Compose(base + [
    RandCropByPosNegLabeld(keys=["image", "label"], label_key="label",
                           spatial_size=PATCH, pos=1, neg=1, num_samples=2,
                           image_key="image", image_threshold=0),
    CastToTyped(keys=["image", "label"], dtype=[torch.float32, torch.float32]),
    RandFlipd(keys=["image", "label"], prob=0.5, spatial_axis=0),
    RandFlipd(keys=["image", "label"], prob=0.5, spatial_axis=1),
    RandRotate90d(keys=["image", "label"], prob=0.3, max_k=3, spatial_axes=(0, 1)),
    RandShiftIntensityd(keys="image", offsets=0.05, prob=0.3),
    EnsureTyped(keys=["image", "label"]),
])
val_tf = Compose(base + [EnsureTyped(keys=["image", "label"])])

def make_train_loader(files):
    ds = PersistentDataset(files, train_tf, cache_dir=os.path.join(CACHE_DIR, "train"))
    return DataLoader(ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=2, pin_memory=True)

def make_val_loader(files):
    ds = PersistentDataset(files, val_tf, cache_dir=os.path.join(CACHE_DIR, "val"))
    return DataLoader(ds, batch_size=1, shuffle=False, num_workers=2)
```

```python
def make_model():
    return UNet(spatial_dims=3, in_channels=1, out_channels=2,
                channels=(16, 32, 64, 128, 256), strides=(2, 2, 2, 2),
                num_res_units=2, norm=Norm.INSTANCE).to(device)

loss_fn = DiceCELoss(to_onehot_y=True, softmax=True, include_background=False)
post_pred = AsDiscrete(argmax=True, to_onehot=2)
post_label = AsDiscrete(to_onehot=2)
dice_metric = DiceMetric(include_background=False, reduction="mean")

def train_one_epoch(model, loader, optimizer, scaler):
    model.train()
    losses = []
    for b in loader:
        x = b["image"].to(device, non_blocking=True)
        y = b["label"].to(device, non_blocking=True)
        optimizer.zero_grad(set_to_none=True)
        with torch.autocast(device_type="cuda", dtype=torch.float16, enabled=use_amp):
            loss = loss_fn(model(x), y)
        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()
        losses.append(loss.item())
    return float(np.mean(losses))

def validate(model, loader):
    """Returns (mean Dice over cases, list of per-case Dice)."""
    model.eval()
    scores = []
    with torch.no_grad():
        for vb in loader:
            vx = vb["image"].to(device).float()
            vy = vb["label"].to(device).float()
            with torch.autocast(device_type="cuda", dtype=torch.float16, enabled=use_amp):
                vo = sliding_window_inference(vx, PATCH, 4, model, overlap=0.25)
            preds = [post_pred(i) for i in decollate_batch(vo)]
            gts = [post_label(i) for i in decollate_batch(vy)]
            dice_metric.reset()
            dice_metric(y_pred=preds, y=gts)
            scores.append(float(dice_metric.aggregate().item()))
    return float(np.nanmean(scores)), scores
```

```python
over_files = train_files[:2]
over_loader = make_train_loader(over_files * 50)         # 100 items = 50 steps per epoch
over_val = make_val_loader(over_files)

m_o = make_model()
opt_o = torch.optim.AdamW(m_o.parameters(), lr=LR, weight_decay=1e-5)
sc_o = torch.amp.GradScaler("cuda", enabled=use_amp)

t0 = time.time()
for e in range(OVERFIT_EPOCHS):
    l = train_one_epoch(m_o, over_loader, opt_o, sc_o)
    d, per = validate(m_o, over_val)
    print(f"overfit epoch {e+1}/{OVERFIT_EPOCHS} | loss {l:.4f} | Dice on the 2 cases {d:.4f} | {(time.time()-t0)/60:.1f} min")
del m_o, opt_o
torch.cuda.empty_cache()
```

```python
train_loader = make_train_loader(train_files)
val_loader_quick = make_val_loader(val_files[:VAL_CASES_DURING])

model = make_model()
optimizer = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=1e-5)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS_TOTAL)
scaler = torch.amp.GradScaler("cuda", enabled=use_amp)

start_epoch, best_dice, best_epoch = 0, -1.0, 0
history = {"loss": [], "val_epoch": [], "val_dice": []}

if os.path.exists(CKPT_LAST):
    ck = torch.load(CKPT_LAST, map_location=device)
    model.load_state_dict(ck["model"]); optimizer.load_state_dict(ck["optimizer"])
    scheduler.load_state_dict(ck["scheduler"]); scaler.load_state_dict(ck["scaler"])
    start_epoch, best_dice, best_epoch, history = ck["epoch"], ck["best_dice"], ck["best_epoch"], ck["history"]
    print(f"Resumed from epoch {start_epoch} (best Dice so far {best_dice:.4f} at epoch {best_epoch})")
else:
    print("Starting from scratch")
```

```python
t_start = time.time()
for epoch in range(start_epoch, EPOCHS_TOTAL):
    t_ep = time.time()
    loss = train_one_epoch(model, train_loader, optimizer, scaler)
    scheduler.step()
    history["loss"].append(loss)
    msg = f"epoch {epoch+1}/{EPOCHS_TOTAL} | loss {loss:.4f} | lr {scheduler.get_last_lr()[0]:.2e} | {(time.time()-t_ep)/60:.1f} min"

    if (epoch + 1) % VAL_EVERY == 0 or epoch + 1 == EPOCHS_TOTAL:
        d, _ = validate(model, val_loader_quick)
        history["val_epoch"].append(epoch + 1)
        history["val_dice"].append(d)
        msg += f" | val Dice ({VAL_CASES_DURING} cases) {d:.4f}"
        if d > best_dice:
            best_dice, best_epoch = d, epoch + 1
            torch.save({"model": model.state_dict(), "epoch": epoch + 1, "dice": d}, CKPT_BEST)
            msg += "  <- best"
    print(msg)

    torch.save({"model": model.state_dict(), "optimizer": optimizer.state_dict(),
                "scheduler": scheduler.state_dict(), "scaler": scaler.state_dict(),
                "epoch": epoch + 1, "best_dice": best_dice, "best_epoch": best_epoch,
                "history": history}, CKPT_LAST)

    if (time.time() - t_start) / 60 > TIME_BUDGET_MIN and epoch + 1 < EPOCHS_TOTAL:
        print(f"Time budget reached after epoch {epoch+1}. Run this cell again to continue.")
        break
else:
    print("Training finished.")
```
